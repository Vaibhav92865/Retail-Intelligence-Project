```python
import pandas as pd 
import numpy as np 
import os 
```

```python
df = pd.read_excel("../data/raw/online Retail.xlsx.xlsx")
print("original Rows:", len(df))
```

```python
df_clean = df.copy()
```

```python
df_clean = df_clean.drop_duplicates()
print("After removing duplicates:", len(df_clean))
```

```python
df_clean = df_clean[
    df_clean["InvoiceNo"].astype(str).str.startswith("C")].copy()
print("After removing cancelled invoices:", len(df_clean))
```

```python
df_clean = df_clean[df_clean["UnitPrice"] > 0].copy()
print("After removing invalid price:", len(df_clean))
```

```python
df_clean = df_clean[df_clean["Quantity"] > 0].copy()
print("After removing negative/zero quantity:", len(df_clean))
```

```python
df_clean["InvoiceDate"] = pd.to_datetime(
df_clean["InvoiceDate"], errors="coerce")
df_clean = df_clean.dropna(subset=["InvoiceDate"]).copy()
print("After date cleaning:", len(df_clean))
```

```python
df_clean["Revenue"] = (
df_clean["Quantity"] * df_clean["UnitPrice"])
```

```python
print("\nFINAL CLEANED DATASET")
print("Rows:", df_clean.shape[0])
print("Columns:", df_clean.shape[1])
print("Missing CustomerID:", df_clean["CustomerID"].isnull().sum())
print("Duplicate Rows:", df_clean.duplicated().sum())
print("Total Revenue:", round(df_clean["Revenue"].sum(), 2))
```

```python
os.makedirs("../data/processed", exist_ok=True)
df_clean.to_csv(
"../data/processed/retail_cleaned.csv",index=False)
```
